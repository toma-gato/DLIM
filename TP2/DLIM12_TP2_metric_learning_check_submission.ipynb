```python
import numpy as np
import os

def check_submission(submission_path: str) -> None:
    expected_shape = (400,)
    expected_dtype = bool
    # Load the submission: if it's a .npy file, use np.load, if it's a .npz file, use np.load and extract the first item
    submission = np.load(submission_path)
    if isinstance(submission, np.lib.npyio.NpzFile):
        submission = submission[list(submission.keys())[0]]
    # Try to reshape it to the expected shape, and fail if it doesn't match
    try:
        submission = submission.reshape(expected_shape)
    except ValueError:
        raise ValueError(f"Invalid shape for submission {submission_path}: {submission.shape} (expected {expected_shape})")
    # Check the dtype
    if submission.dtype != expected_dtype:
        raise ValueError(f"Invalid dtype for submission {submission_path}: {submission.dtype} (expected {expected_dtype})")

    # Notify the user that the submission is valid
    print(f"Submission {submission_path} is valid. Make sure you check the results visually before submitting them.")
```
